```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 30],
    'Department': ["HR", "Finance", "Finance", np.nan, 'IT']
})

print("Original Dataset (With Missing Values):")
print(df)

# Fill missing Age with mean and Department with mode
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])

print("\nDataset After Handling Missing Values:")
print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 30],
    'Department': ["HR", "Finance", "Finance", np.nan, 'IT']
})

print("Original Dataset:")
print(df)

df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print("\nAfter Forward Fill:\n", df_ffill)

df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print("\nAfter Backward Fill:\n", df_bfill)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 30],
    'Department': ["HR", "Finance", "Finance", 'IT', 'IT']
})

print("Original Dataset:\n", df)

df_drop_rows = df.dropna()
print("\nAfter Dropping The Rows:\n", df_drop_rows)

df_drop_cols = df.dropna(axis=1)
print("\nAfter Dropping The Columns:\n", df_drop_cols)
```

```python
import pandas as pd

# Exact match duplicate removal dataset
df_exact_data = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
})
print("Original Data:\n", df_exact_data)
print("\nAfter Exact Match Removal:\n", df_exact_data.drop_duplicates())

# Subset-based duplicate removal
df_duplicates = pd.DataFrame({
    'ID': [1, 2, 2, 3, 4, 4],
    'Name': ['Alice', 'Bob', 'Bob', 'Charlie', 'David', 'David'],
    'Age': [25, 30, 30, 35, 40, 40]
})

print("\nOriginal Dataset with Duplicates:\n", df_duplicates)

df_subset_id = df_duplicates.drop_duplicates(subset=["ID"])
print("\nAfter Subset-Based Removal (ID):\n", df_subset_id)

df_subset_name = df_duplicates.drop_duplicates(subset=["Name"])
print("\nAfter Subset-Based Removal (Name):\n", df_subset_name)
```
