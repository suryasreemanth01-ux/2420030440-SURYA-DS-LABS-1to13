```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

# 1. Version Checks and Basics
print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)
print("Scikit-learn Version:", sklearn.__version__)
print("Hello, Data Science!\n")

# 2. NumPy Array & Pandas DataFrame Basics
a = np.array([10, 20, 30, 40])
print("NumPy Array:", a)

data = {
    "Name": ["Ram", "Sita", "John"],
    "Age": [20, 21, 22]
}
df = pd.DataFrame(data)
print("\nPandas DataFrame:")
display(df)

# 3. Matplotlib Basic Plotting
x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# 4. Python Variables, Data Types, and Input/Output
var_a = 171
var_b = "Giridhar"
var_c = 3.33333
print(var_a, var_b, var_c, sep="\n")

print("\nData Types:")
print("Type of a: ", type(var_a))
print("Type of b: ", type(var_b))
print("Type of c: ", type(var_c))

# 5. String Methods and Slicing Examples
a_str = " Hello World "
print("\nOriginal String:", a_str)
print("Upper:", a_str.upper())
print("Lower:", a_str.lower())
print("Title:", a_str.title())
print("Reversed (Slicing):", a_str[::-1])
print("First 5 characters:", a_str[0:5])
```
