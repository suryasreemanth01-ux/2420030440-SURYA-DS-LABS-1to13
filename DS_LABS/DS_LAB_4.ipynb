```python
import matplotlib.pyplot as plt

y1 = []  # stores +ve values
y2 = []  # stores -ve values

x = range(-100, 100, 10)

for i in x:
    y1.append(i**2)

for i in x:
    y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)

plt.xlabel("x")
plt.ylabel("y")

plt.ylim(-2000, 2000)

plt.axhline(0)
plt.axvline(0)

plt.savefig("quad.png")
plt.show()
```

```python
import dash
from dash import dcc, html, Input, Output

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H3("Square Calculator"),

    dcc.Input(
        id="input",
        type="number",
        placeholder="Enter a Number"
    ),

    html.Br(),
    html.Br(),

    html.Div(id="output")
])

@app.callback(
    Output("output", "children"),
    Input("input", "value")
)
def update_value(input_data):
    if input_data is None:
        return "Enter a number"
    return f"Square = {input_data ** 2}"

app.run(debug=False, jupyter_mode="inline")
```
