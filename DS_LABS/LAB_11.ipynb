```python
# Classification using Decision Tree & Performance Evaluation Metrics
# Application: Student Pass/Fail Prediction
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Training Data Features [Study Hours, Attendance Percentage] and Labels (0=Fail, 1=Pass)
X = np.array([
    [2, 50], [3, 60], [4, 65], [5, 70],
    [6, 75], [7, 80], [8, 85], [9, 90]
])
y = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Train Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# Test Data and Actual Outputs
X_test = np.array([
    [3, 55],
    [5, 72],
    [8, 88],
    [4, 60]
])
y_test = np.array([0, 1, 1, 0])

# Predict outputs
y_pred = model.predict(X_test)

# Performance Evaluation
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

# Display Results
print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred)
print("\nConfusion Matrix:\n", cm)
print("Accuracy:", accuracy)
print("Error Rate:", error_rate)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```

```python
# K-Nearest Neighbors (KNN) Classification
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
)

X_train = np.array([
    [2, 50], [3, 60], [4, 65], [5, 70],
    [6, 75], [7, 80], [8, 85], [9, 90]
])
y_train = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Create KNN Model with k = 3
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

X_test = np.array([[3, 55], [5, 72], [8, 88], [4, 60]])
y_test = np.array([0, 1, 1, 0])
y_pred = model.predict(X_test)

# Evaluation
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("Actual Labels:", y_test)
print("Predicted Labels:", y_pred)
print("\nConfusion Matrix:\n", cm)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```
